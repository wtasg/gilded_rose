````python
import numpy as np

def matrix_addition(A, B):
    if A.shape != B.shape:
        print("Different shapes.")
        return np.zeros(1, dtype=A.dtype)

    result = np.empty(A.shape, dtype=A.dtype)

    for i in np.ndindex(A.shape):
        result[i] = A[i] + B[i]

    return result

print(matrix_addition(np.array([[1, 2], [3, 4]]), np.array([[1, 2], [3, 4]])))

A = np.array([[1, 2], [3, 4]])
B = np.array([[1, 2], [3, 4]])
for i in range(A.shape[0]):
    for j in range(A.shape[1]):
        print(A[i][j], B[i][j], A[i][j] + B[i][j])
        
````

````python
%%manim -ql -v ERROR --progress_bar None MatrixTimesVector

from manim import *
import numpy as np


class MatrixTimesVector(LinearTransformationScene):
    def __init__(self, **kwargs):
        super().__init__(
            show_coordinates=True,
            leave_ghost_vectors=True,
            **kwargs,
        )

    def construct(self):
        A = np.array([
            [2, 1],
            [0, 1],
        ])

        x = np.array([2, 1])
        Ax = A @ x

        # Mathematical definitions
        A_math = MathTex(
            r"A = \begin{bmatrix} 2 & 1 \\ 0 & 1 \end{bmatrix}"
        )

        x_math = MathTex(
            r"x = \begin{bmatrix} 2 \\ 1 \end{bmatrix}"
        )

        Ax_math = MathTex(
            r"Ax = \begin{bmatrix} 5 \\ 1 \end{bmatrix}"
        )

        definitions = VGroup(A_math, x_math)
        definitions.arrange(DOWN, aligned_edge=LEFT, buff=0.4)
        definitions.to_corner(UL, buff=0.6)

        self.play(Write(A_math))
        self.play(Write(x_math))
        self.wait()

        # Geometric x
        vector = self.add_vector(
            x,
            color=YELLOW,
        )

        vector_label = MathTex("x")
        vector_label.next_to(vector.get_end(), UR, buff=0.15)

        self.play(Write(vector_label))
        self.wait()

        # Indicate Ax
        self.play(Write(Ax_math))
        self.wait()

        # Apply A to the geometric vector/plane
        self.apply_matrix(A, run_time=3)

        self.wait()
````

## Matrix transformations

### scaling

```text
A = [sx  0
      0 sy]
```

### rotation

By angle θ:

```text
R = [ cosθ  -sinθ
      sinθ   cosθ ]
```

By 90° counter-clockwise:

```text
R = [ 0  -1
      1   0]
```

### reflection

mirror image

across x-axis:

```text
A = [1   0
     0  -1]
```

across y-axis:

```text
A = [-1  0
      0  1]
```

### shear

x shear

```text
A = [1  1
     0  1]
```

y shear

```text
A = [1  0
     1  1]
```

### projection

x-axis projection:

```text
A = [1  0
     0  0]
```

y-axis projection:

```text
A = [0  0
     0  1]
```

### 180 rotation

```text
A = [-1  0
      0  -1]
```
