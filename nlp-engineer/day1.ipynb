````python
# Vector addition
## how to get shape of a given vector?: np.array([1, 2, 3]).shape
### how to get number of elements: np.array([1, 2, 3]).shape[0]

v = np.array([1, 2, 3])
print(v, v.shape, v.shape[0], v.dtype)

## how to loop over a np vector?
for i in range(v.shape[0]):
    print(i)

## how to create an empty vector?
e = np.empty(v.shape, v.dtype)
print(e, e.shape, e.shape[0], e.dtype)

## how to set value in a vector at a given vertex
e[0] = v[0]+1
e[1] = v[1]+1
e[2] = v[2]+1
# e[3] = v[3]+1 # IndexError: index 3 is out of bounds for axis 0 with size 3
print(e)

## full function

def vector_addition_1(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shape vectors cannot be added atm.")
        return np.zeros(1, dtype="int64")

    return np.ones(1)

print(vector_addition_1(np.array([1, 2, 3]), np.array([[1, 2], [4, 5]]))) 
# Different shape vectors cannot be added atm.
# [0]

def vector_addition_2(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shape vectors cannot be added atm.")
        return np.zeros(1, dtype="int64")

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data typed vectors cannot be added atm.")
        return np.zeros(1, dtype=udt)

    return np.ones(1)

print(vector_addition_2(np.array([1, 2, 3]), np.array([1.0, 2.0, 3.0]))) 
# Different data typed vectors cannot be added atm.
# [0]

def vector_addition_3(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shape vectors cannot be added atm.")
        return np.zeros(1, dtype="int64")

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data typed vectors cannot be added atm.")
        return np.zeros(1, dtype=udt)

    e = np.empty(ushp, udt)
    for i in range(ushp[0]):
        e[i] = u[i] + v[i]
        
    return e

print(vector_addition_3(np.array([1, 2, 3]), np.array([1.0, 2.0, 3.0], dtype="int64"))) 
# [2 4 6]
print(vector_addition_3(np.array([1, 2, 3]), np.array([-1, -2, -3]))) 
# [0 0 0]
````

````python
def vector_addition(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shape vectors cannot be added atm.")
        return np.zeros(1, dtype="int64")

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data typed vectors cannot be added atm.")
        return np.zeros(1, dtype=udt)

    e = np.empty(ushp, udt)
    for i in np.ndindex(ushp):
        e[i] = u[i] + v[i]
        
    return e


print(vector_addition(np.array([1, 2, 3]), np.array([1.0, 2.0, 3.0], dtype="int64"))) 
# [2 4 6]
print(vector_addition(np.array([1, 2, 3]), np.array([-1, -2, -3]))) 
# [0 0 0]
print(vector_addition(np.array([[1, 2], [3, 4]]), np.array([[1, 2], [3, 4]])))
# [[2 4]
#  [6 8]]
## Glory to np.ndindex which takes care of proper indexing. Yayiya!
````

````python
def vector_subtraction(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shape vectors cannot be subtracted atm.")
        return np.zeros(1, dtype="int64")

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data typed vectors cannot be subtracted atm.")
        return np.zeros(1, dtype=udt)

    e = np.empty(ushp, udt)
    for i in np.ndindex(ushp):
        e[i] = u[i] - v[i]
        
    return e

print(vector_subtraction(np.array([1, 2, 3]), np.array([1.0, 2.0, 3.0], dtype="int64"))) 
# [0 0 0]
print(vector_subtraction(np.array([1, 2, 3]), np.array([-1, -2, -3]))) 
# [2 4 6]
print(vector_subtraction(np.array([[1, 2], [3, 4]]), np.array([[1, 2], [3, 4]])))
# [[0 0]
#  [0 0]]
````

````python
def vector_dot_product_1(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shapes.")
        return -1

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data types.")
        return -1

    out = 0
    for i in range(ushp[0]):
        out += u[i]*v[i]

    return out

print(vector_dot_product_1(np.array([1, 2, 3]), np.array([4, 5, 6])))
# 32

def vector_dot_product(u, v):
    ushp, vshp = u.shape, v.shape
    if ushp != vshp:
        print("Different shapes.")
        return -1

    udt, vdt = u.dtype, v.dtype
    if udt != vdt:
        print("Different data types.")
        return -1

    out = 0
    for i in np.ndindex(ushp):
        out += u[i]*v[i]

    return out

print(vector_dot_product(np.array([1, 2, 3]), np.array([4, 5, 6])))
# 32
print(vector_dot_product(np.array([[1, 2], [3, 4]]), np.array([[1, 2], [3, 4]])))
````
